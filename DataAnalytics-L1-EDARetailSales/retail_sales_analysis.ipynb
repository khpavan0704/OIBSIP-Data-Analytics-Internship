```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df.info()
```

```python
df['Month'] = df['Date'].dt.to_period('M')
```

```python
monthly_sales = df.groupby('Month')['Total Amount'].sum()
monthly_sales
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

### Observation – Monthly Sales Trend

- May 2023 recorded the highest monthly sales.
- September 2023 recorded one of the lowest monthly sales.
- Sales increased significantly from September to October 2023.
- December 2023 also showed strong sales performance.
- January 2024 shows very low sales, which may be because the dataset contains only a limited number of transactions for that month.

```python
df['Quarter'] = df['Date'].dt.to_period('Q')
```

```python
quarterly_sales = df.groupby('Quarter')['Total Amount'].sum()
quarterly_sales
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

quarterly_sales.plot(kind='bar')

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.grid(axis='y')

plt.show()
```

### Observation – Quarterly Sales Trend

- 2023 Q4 recorded the highest quarterly sales.
- 2023 Q2 recorded the second-highest quarterly sales.
- 2023 Q3 recorded the lowest quarterly sales among the full quarters of 2023.
- Sales increased from Q3 to Q4 in 2023.
- 2024 Q1 shows very low sales because the dataset contains only January 2024 transactions.

```python
df['Age Group'] = pd.cut(
    df['Age'],
    bins=[17, 25, 35, 45, 55, 65],
    labels=['18-25', '26-35', '36-45', '46-55', '56-65']
)
```

```python
age_group_counts = df['Age Group'].value_counts().sort_index()
age_group_counts
```

```python
plt.figure(figsize=(10, 6))

age_group_counts.plot(kind='bar')

plt.title('Customer Distribution by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.grid(axis='y')

plt.show()
```

```python
gender_counts = df['Gender'].value_counts()
gender_counts
```

```python
plt.figure(figsize=(8, 6))

gender_counts.plot(kind='bar')

plt.title('Customer Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.grid(axis='y')

plt.show()
```

### Observation – Customer Age Groups

- Customers are distributed across all five age groups.
- The 46–55 age group has the highest number of customers.
- The 18–25 age group has the lowest number of customers.
- The 26–35 and 36–45 age groups have relatively similar customer counts.
- The customer base is spread across a wide range of age groups.

```python
category_sales = df.groupby('Product Category')['Total Amount'].sum().sort_values(ascending=False)

category_sales
```

```python
plt.figure(figsize=(8, 6))

category_sales.plot(kind='bar')

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=0)
plt.grid(axis='y')

plt.show()
```

### Observation – Revenue by Product Category

- Electronics generated the highest total revenue.
- Clothing generated the second-highest total revenue.
- Beauty generated the lowest total revenue among the three categories.
- The revenue difference between the categories is relatively small.

```python
correlation = df[['Age', 'Quantity', 'Price per Unit', 'Total Amount']].corr()

correlation
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='Product Category',
    y='Total Amount'
)

plt.title('Distribution of Transaction Amount by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Amount')

plt.show()
```

### Observation – Transaction Amount Distribution

- The box plot shows the distribution of transaction amounts across product categories.
- It helps compare the typical transaction value and variability between categories.
- Some transactions have substantially higher values than the typical transaction amount.
- The distribution can help identify unusual high-value transactions that may require further investigation.

## Business Recommendations

### 1. Focus on Electronics
Electronics generated the highest total revenue among the product categories. 
The business should maintain sufficient inventory for Electronics and consider 
promotional campaigns for high-performing products in this category.

### 2. Target Different Age Groups
Customers are distributed across multiple age groups, with the 46–55 age group 
being the largest. Marketing campaigns can be designed specifically for this 
group while also creating targeted offers to attract younger customers.

### 3. Investigate Monthly Sales Fluctuations
Monthly sales show significant fluctuations, with May 2023 and December 2023 
showing strong performance while September 2023 was relatively low. The business 
should investigate the reasons behind these fluctuations and use successful 
sales strategies from high-performing months during weaker periods.
